# Exploring the simulator
Run cells top to bottom (after `pip install -e .`).

```python
from lob.simulation import SimulationConfig, run_simulation
from lob.visualization import charts

run = run_simulation(SimulationConfig(n_events=20_000, seed=42), snapshot_every=50)
snap, ex = run.snapshots_df(), run.executions_df()
snap.tail()
```

## Mid price, spread, imbalance over time

```python
charts.price_panel(snap)
```

## Final book

```python
charts.depth_chart(run.engine.book)
```

## Does bigger mean costlier? (slippage & impact vs order size)

```python
ex.groupby(ex["quantity"].round(-1).clip(upper=200))[["slippage_bps", "impact_bps"]].mean()
```

```python
charts.execution_scatter(ex)
```

## Experiment
Change `price_sigma_ticks`, `p_market`, or `arrival_rate` and re-run. How does the average spread change? Why?

```python
for sigma in (2, 8, 20):
    r = run_simulation(SimulationConfig(n_events=20_000, price_sigma_ticks=sigma), snapshot_every=50)
    print(sigma, r.snapshots_df()["spread"].mean(), r.executions_df()["slippage_bps"].mean())
```
